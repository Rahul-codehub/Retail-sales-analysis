```python
# Retail Sales Data Analysis & Business Insights

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Retail Sales Data Analysis")
```

```python
import os

print(os.getcwd())
```

```python
# Load the cleaned dataset

df = pd.read_csv("data/cleaned_superstore_dataset.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
# Basic dataset information

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Check data quality

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
# Overall business performance

total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_quantity = df["quantity"].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
```

```python
# Sales and profit by category

category_summary = (
    df.groupby("category")[["sales", "profit"]]
    .sum()
    .sort_values("sales", ascending=False)
)

category_summary
```

```python
# Sales and profit by region

region_summary = (
    df.groupby("region")[["sales", "profit"]]
    .sum()
    .sort_values("sales", ascending=False)
)

region_summary
```

```python
# Sales and profit by customer segment

segment_summary = (
    df.groupby("segment")[["sales", "profit"]]
    .sum()
    .sort_values("sales", ascending=False)
)

segment_summary
```

```python
# Sales and profit by subcategory

subcategory_summary = (
    df.groupby("subcategory")[["sales", "profit"]]
    .sum()
    .sort_values("sales", ascending=False)
)

subcategory_summary
```

```python
# Yearly sales and profit analysis

yearly_summary = (
    df.groupby("order_years")[["sales", "profit"]]
    .sum()
)

yearly_summary
```

```python
# Monthly sales analysis

monthly_sales = df.groupby("order_month")["sales"].sum()

print("Monthly Sales:")
print(monthly_sales)

print("\nBest Sales Month:", monthly_sales.idxmax())
print("Best Month Sales:", round(monthly_sales.max(), 2))
```

```python
# Discount impact analysis

discount_summary = (
    df.groupby("discount")
    .agg(
        sales=("sales", "sum"),
        profit=("profit", "sum"),
        transactions=("order_id", "count")
    )
    .sort_index()
)

discount_summary
```

```python
# Top and bottom subcategories by profit

top_profit_subcategories = subcategory_summary.sort_values(
    "profit", ascending=False
).head(5)

bottom_profit_subcategories = subcategory_summary.sort_values(
    "profit", ascending=True
).head(5)

print("Top 5 Subcategories by Profit:")
print(top_profit_subcategories)

print("\nBottom 5 Subcategories by Profit:")
print(bottom_profit_subcategories)
```

```python
# Visualization 1: Sales by Category

category_summary["sales"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Profit by Region

region_summary["profit"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Sales by Customer Segment

segment_summary["sales"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Monthly Sales Trend

monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Visualization 5: Discount vs Average Profit

avg_profit_by_discount = df.groupby("discount")["profit"].mean()

avg_profit_by_discount.plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Discount vs Average Profit")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.tight_layout()
plt.show()
```

```python
# Visualization 6: Top 10 Subcategories by Sales

top10_subcategories = subcategory_summary.head(10)

top10_subcategories["sales"].sort_values().plot(
    kind="barh",
    figsize=(9, 6)
)

plt.title("Top 10 Subcategories by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Subcategory")
plt.tight_layout()
plt.show()
```

```python
# Key Business Insights

print("KEY BUSINESS INSIGHTS")
print("=" * 50)

print(f"1. Total Sales: {total_sales:,.2f}")
print(f"2. Total Profit: {total_profit:,.2f}")
print(f"3. Total Quantity Sold: {total_quantity:,}")

best_category = category_summary["sales"].idxmax()
best_region = region_summary["profit"].idxmax()
best_segment = segment_summary["sales"].idxmax()
best_subcategory = subcategory_summary["sales"].idxmax()

print(f"4. Highest Sales Category: {best_category}")
print(f"5. Most Profitable Region: {best_region}")
print(f"6. Highest Sales Customer Segment: {best_segment}")
print(f"7. Highest Sales Subcategory: {best_subcategory}")
print(f"8. Best Sales Month: {monthly_sales.idxmax()}")
```

```python
from docx import Document
import pandas as pd

# Load cleaned dataset
df = pd.read_csv("data/cleaned_superstore_dataset.csv")

# Calculate key metrics
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_quantity = df["quantity"].sum()

# Create report
doc = Document()

# Title
title = doc.add_heading(
    "Retail Sales Data Analysis & Business Insights", 0
)
title.alignment = 1

subtitle = doc.add_paragraph("Business Recommendation Report")
subtitle.alignment = 1

doc.add_paragraph("Dataset: Superstore Retail Sales Dataset")
doc.add_paragraph(f"Records analyzed: {len(df):,}")

# Executive Summary
doc.add_heading("1. Executive Summary", level=1)
doc.add_paragraph(
    "This report analyzes retail sales data to identify patterns in "
    "sales, profit, customer segments, regions, product categories, "
    "subcategories, and discounts. The objective is to identify key "
    "business drivers and provide actionable recommendations."
)

# Key Performance
doc.add_heading("2. Key Business Performance", level=1)
doc.add_paragraph(f"Total Sales: ${total_sales:,.2f}")
doc.add_paragraph(f"Total Profit: ${total_profit:,.2f}")
doc.add_paragraph(f"Total Quantity Sold: {total_quantity:,}")

# Category Insights
doc.add_heading("3. Category Insights", level=1)
doc.add_paragraph(
    "Technology was the strongest product category, generating "
    "$836,154.03 in sales and $145,454.95 in profit."
)
doc.add_paragraph(
    "Furniture generated $741,718.42 in sales but only $18,463.33 "
    "in profit, indicating relatively weak profitability."
)
doc.add_paragraph(
    "Office Supplies generated $719,047.03 in sales and "
    "$122,490.80 in profit."
)

# Regional Insights
doc.add_heading("4. Regional Insights", level=1)
doc.add_paragraph(
    "The West region generated the highest sales at $725,457.82 "
    "and the highest profit at $108,418.45."
)
doc.add_paragraph(
    "The Central region generated $501,239.89 in sales but only "
    "$39,706.36 in profit, suggesting opportunities for better "
    "pricing, product mix, or cost management."
)

# Customer Segment Insights
doc.add_heading("5. Customer Segment Insights", level=1)
doc.add_paragraph(
    "The Consumer segment was the largest contributor, generating "
    "$1,161,401 in sales and $134,119.21 in profit."
)
doc.add_paragraph(
    "Corporate customers generated approximately $706,146 in sales "
    "and $91,979.13 in profit."
)
doc.add_paragraph(
    "Home Office generated approximately $429,372 in sales and "
    "$60,310.74 in profit."
)

# Sales Trend
doc.add_heading("6. Sales Trend Insights", level=1)
doc.add_paragraph(
    "Sales increased from $483,966.13 in 2019 to $733,215.26 in "
    "2022. Profit also increased from $49,556.03 in 2019 to "
    "$93,439.27 in 2022, showing strong overall business growth."
)
doc.add_paragraph(
    "November was the highest-sales month based on the monthly analysis."
)

# Product Insights
doc.add_heading("7. Product and Subcategory Insights", level=1)
doc.add_paragraph(
    "Copiers were the most profitable subcategory, generating "
    "$55,617.82 in profit. Phones, Accessories, Paper, and Binders "
    "were also among the strongest contributors."
)
doc.add_paragraph(
    "Tables generated a loss of $17,725.48 despite producing "
    "$206,965.53 in sales. Bookcases also generated a loss of "
    "$3,472.56."
)

# Recommendations
doc.add_heading("8. Business Recommendations", level=1)

recommendations = [
    "Increase marketing and inventory focus on Technology products "
    "because the category leads both sales and profit.",
    "Review Furniture pricing, discounting, and product costs because "
    "it generates high sales but comparatively low profit.",
    "Investigate Tables and Bookcases to determine the causes of "
    "negative profitability and consider pricing or product-mix changes.",
    "Use the West region as a performance benchmark and develop "
    "improvement strategies for lower-performing regions.",
    "Continue targeting the Consumer segment while developing "
    "strategies to increase Corporate and Home Office customer value.",
    "Use monthly sales patterns to improve inventory planning and "
    "promotional campaigns, particularly around November.",
    "Monitor discount levels carefully because excessive discounting "
    "can reduce profitability."
]

for i, recommendation in enumerate(recommendations, 1):
    doc.add_paragraph(f"{i}. {recommendation}")

# Conclusion
doc.add_heading("9. Conclusion", level=1)
doc.add_paragraph(
    "The analysis demonstrates strong overall sales and profit growth. "
    "Technology, the West region, and the Consumer segment are major "
    "business performance drivers. However, Furniture has relatively "
    "low profitability, while Tables and Bookcases generate losses. "
    "Targeted pricing, inventory management, discount control, and "
    "product-level optimization can help improve profitability while "
    "supporting continued sales growth."
)

# Save report
report_path = "report/retail_business_recommendation_report.docx"
doc.save(report_path)

print("Report created successfully!")
print(report_path)
```
